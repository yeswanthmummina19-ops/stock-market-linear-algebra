```python
def solve_two_sum(nums, target):
    # 'len(nums)' is how many numbers are in the list
    for i in range(len(nums)):
        # Look at the numbers AFTER the current one
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
```

```python
my_numbers = [2, 7, 11, 15]
my_target = 13
result = solve_two_sum(my_numbers, my_target)

print(f"The positions are: {result}")
```
